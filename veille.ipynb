### E. La classification supervisée

- Le classement automatique ou **classification supervisée** sert à catégoriser des données étiquetées à partir d'algorithme et/ou réseau de neurones. </br>
- En se basant sur des données statistiques, il attribue une classe à un objet/humain. </br>
- La classification supervisée se fait grâce à l'apprentissage supervisée.

**Son objectif =** Prédire, classer, catégoriser les données étiquetées </br>
**Type de donnée utilisée =** Données étiquetées. </br>
**Exemple de cas concret =** Plusieurs images d'animaux, avec une étiquette donnant l'espèce de chaque animal constituent sa base de données. Puis en donnant une nouvelle image, il doit deviner l'espèce de celui-ci. </br>

### F. La classification non supervisée

- La **classification non supervisée** utilise des données non étiquetées pour s'entraîner (apprentissage non supervisée) et classifie les données en fonction de ce qu'il a compris. </br>
- Elle fonctionne indépendamment pour apprendre la structure naturel des données, sans conseils ni instructions spécifiques. </br>
- Il suffit de fournir les données d'entrée et de laisser l'algorithme identifier tout modèle déjà présent dans l'ensemble de données. </br>

**Son objectif =** Organiser, segmenter ou simplifier les données, sans connaissance a priori. </br>
**Type de donnée utilisée =** Données non étiquetées. </br>
**Exemple de cas concret =** Pleins de fruits et légumes composent sa base de donnée, et en se basant sur son algorithme il va deviner, sans connaissance au préalable, comment catégoriser tous ces aliments. </br>

### G. La régression

- Méthode statistique utilisée pour trouver la relation entre les variables et qui permet de prédire des résultats en fonction d'un ensemble de variables d'entrée.
- Dans un contexte de ML, la régression linéaire trouve la relation entre les caractéristiques et un libellé.
- Elle est matérialisée par un graphique et se base sur une table de donnée.

**Son objectif =** Prévoir la valeur de données inconnues en utilisant une autre valeur de données apparentée et connue. </br>
**Type de donnée utilisée =** Nombres/Chiffres</br>

### H. La validation croisée

- Méthodes de rééchantillonnage et de division d'échantillons qui utilisent différentes portions des données pour tester et entraîner un modèle.
- Elle est souvent employée dans des contextes où l'objectif est la prédiction, et où l'on souhaite estimer la précision pratique d'un modèle prédictif.
- Il est nécessaire d’écarter en amont une partie des données du dataset d’entraînement. Ces données seront utilisées pour tester et valider le modèle uniquement.
- Les données : + variable/diversité = plus le modèle apprend mieux à fonctionner.

**Son objectif =** Tester les performances et la précision d'un modèle prédictif de ML à généraliser. Régler les hyperparamètres d'un modèle. Comparer plusieurs modèles pour sélectionner celui le plus approprié.</br>
**Type de donnée utilisée =** Modèles entraînés savec des données étiquetées ainsi que non étiquetées.</br>

*A savoir : Si le modèle est peu performant sur différents ensembles de données de test mais performant sur les données d'entraînement lors des tests, cela indique un surapprentissage.*
