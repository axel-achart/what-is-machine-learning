## I. Les données d’entraînement, de test et/ou de validation

### Pour entraîner un modèle de machine learning sur un jeu de données, il faut s'assurer que notre modèle n'apprend pas les résultats par cœur.
Les résultats de test sur des exemples différents sont plus solides que des preuves sur un seul ensemble de données.
Pour obtenir ces différents exemples, on sépare le dataset en deux (souvent 80/20) :

- Un ensemble d'entraînement pour entraîner le modèle
- Un ensemble de test pour évaluer le modèle

![training set .png](<attachment:training set .png>)

### Cette approche n'est pas optimale. Supposons que l'on entraîne un modèle sur le "training set" et que l'on évalue sur l'ensemble test à plusieurs reprises.
À chaque tour, si on utilise l'ensemble de test pour ajuster les hyperparamètres et les fonctionnalités,
le modèle risque de s'adapter étroitement à l'ensemble de test (surapprentissage).

![training set avec validation .png](<attachment:training set avec validation .png>)

Il faut ajouter un ensemble de validation qui sera utilisé pour évaluer l'ensemble d'entraînement.
Une fois l'ensemble d'entraînement validé par le validation set, on peut utiliser le test set pour vérifier le modèle.

![workflow trainning set .png](<attachment:workflow trainning set .png>)

Un bon ensemble d'entraînement :
- Suffisamment de données pour générer des résultats de test statistiquement pertinents
- Représentatif de l'ensemble de données global. Autrement dit, ne choisissez pas un ensemble de test dont les caractéristiques sont différentes de celles de l'ensemble d'entraînement.
- Représentatif des données réelles que le modèle rencontrera une fois validé
- Aucun exemple en double dans le modèle d'entraînement

## J. Corrélation linéaire (de Pearson) entre deux variables

### La corrélation de Pearson examine la relation entre deux variables.

### Force de corrélation

0,0 < 0,1 : Pas de corrélation
0,1 < 0,3 : Faible corrélation
0,3 < 0,5 : Corrélation moyenne
0,5 < 0,7 : Corrélation élevée
0,7 < 1 : Corrélation très élevée

Une corrélation est dite positive lorsque les valeurs d'une variable sont associées aux valeurs d'une autre variable, donc les deux évoluent en même temps et dans le même sens.
Une corrélation est dite négative lorsque les valeurs d'une variable évoluent en même temps mais dans des sens opposés.

### Calculer la corrélation de Pearson

### Hypothèses dans l'analyse de corrélation de Pearson

*Hypothèse nulle* : Le coefficient de corrélation de Pearson n'est pas significativement différent de 0 → il n'y a pas de relation linéaire.
*Hypothèse alternative* : Le coefficient de corrélation s'écarte significativement de 0 → il existe une corrélation linéaire.

### Significativité et test t

La question de savoir si le coefficient de corrélation de Pearson est significativement différent de zéro sur la base de l'échantillon étudié peut être vérifiée à l'aide d'un test t. Ici, r est le coefficient de corrélation et n la taille de l'échantillon.

Une valeur p peut alors être calculée à partir de la statistique de test t. Si la valeur p est inférieure au seuil de signification spécifié, généralement 5 %, l'hypothèse nulle est rejetée, sinon elle ne l'est pas.

#### Conditions de la corrélation de Pearson
Mais qu'en est-il des prérequis pour une corrélation de Pearson ? Il faut distinguer si l'on veut simplement calculer le coefficient de corrélation de Pearson ou tester une hypothèse.

Pour calculer le coefficient de corrélation de Pearson, il faut seulement deux variables métriques. Les variables métriques sont, par exemple, le poids d'une personne, son salaire ou sa consommation d'électricité.

Le coefficient de corrélation de Pearson nous indique alors l'importance de la relation linéaire. S'il existe une corrélation non linéaire, nous ne pouvons pas la déduire du coefficient de corrélation de Pearson.

## K. Une fonction de coût

#### La fonction de coût permet de mesurer l'écart entre les prédictions d'un modèle et les valeurs réelles.

### L'objectif est donc de minimiser cette fonction de coût.

Il existe plusieurs fonctions de coût.
Chacune répond à un problème spécifique et aux caractéristiques des données.

### Les principales fonctions de coût sont :
- La moyenne des erreurs quadratiques (MSE)
- L'erreur absolue moyenne (MAE)
- Log Loss (Logarithme de perte)

#### La moyenne des erreurs quadratiques (MSE)

Elle est principalement utilisée pour des modèles de régression.
Cette méthode des moindres carrés pénalise fortement les grandes erreurs en les élevant au carré.
**Utile pour des prédictions où les écarts importants sont critiques.**

![F MSE.png](<attachment:F MSE.png>)

J(θ) : la valeur de la fonction de coût (ou « erreur moyenne quadratique ») pour un ensemble donné de paramètres θ

m : le nombre total d’exemples dans le dataset.
hθ(x(i)) : la prédiction du modèle pour l’exemple i, souvent appelée « hypothèse ».

y(i) : la valeur réelle attendue pour l’exemple i.

(hθ(x(i))−y(i))² : le carré de l’erreur entre la prédiction et la vraie valeur pour l’exemple i.

#### L'erreur absolue moyenne (MAE)

Il s’agit d’une autre formule pour les modèles de régression linéaire. Mais à la différence de la précédente, cette fonction de coût mesure la moyenne des erreurs absolues entre les valeurs réelles et les prédictions d’un modèle. Et comme la MAE n’élève pas les erreurs au carré, elle est beaucoup plus robuste face aux valeurs aberrantes.

### Log Loss (algorithme de perte)

Cette fonction de coût est utilisée pour évaluer les modèles de classification probabiliste, notamment dans les problèmes de classification binaire ou multiclasse.

Le Log Loss mesure la distance entre les distributions des probabilités prédites et les vraies classes.

## L. La descente de gradient

### La descente de gradient est l'algorithme "moteur" qui permet au modèle de réduire son erreur. Si la fonction de coût est une montagne, la descente de gradient est la méthode pour trouver le point le plus bas (la vallée) en marchant dans le brouillard.

##### La descente de gradient est un algorithme d’optimisation couramment utilisé pour entraîner les modèles de machine learning et les réseaux neuronaux. Ce type d’algorithme entraîne les modèles de machine learning par réduction des erreurs entre les résultats prédits et les résultats réels.

### Comment ça marche ?

L'idée est de trouver le minimum d'une fonction mathématique sans la résoudre de manière traditionnelle.

On met en place un algorithme itératif qui avance pas à pas pour s'approcher du minimum.

L'équation correspond à la relation entre les données.

### Description de l'algorithme

![DESCENTE DE GRADIANT.png](<attachment:DESCENTE DE GRADIANT.png>)

##### Nous avons les deux fonctions (dérivées) dont nous allons chercher la valeur la plus petite (pour a et b). Pour cela, nous allons parcourir ces courbes (l’une et l’autre) pas à pas, comme notre homme en haut de sa montagne.

Chaque pas (epoch) est guidé par le learning rate, qui est en quelque sorte la longueur du pas. On a donc plusieurs pas à faire jusqu’à trouver notre minimum (cf. escalier en rouge dans la figure ci-dessous).

### Deux remarques :

- Il faut bien commencer quelque part, donc on initialisera par des valeurs au hasard a (a0) et b (b0)
- Attention à ne pas choisir un learning rate trop faible, sinon vous allez mettre beaucoup de temps à arriver au minimum, ou trop grand au risque de le manquer !

Nous allons donc avancer pas à pas, et à chaque pas, nous allons calculer les deux gradients (de a et b). Une fois que l’on aura calculé ces gradients, nous aurons deux indications :

- Le sens dans lequel nous devons aller (le signe du gradient). Si notre gradient est positif, cela signifie que la pente monte vers la droite, on doit donc aller vers la gauche, et vice versa. Pour résumer, on doit toujours aller du côté opposé à la pente.
- L’amplitude du pas (proportionnelle à la valeur du gradient). Plus la pente est forte, plus l’amplitude du pas sera grande ! Plus on va se rapprocher d’une pente nulle (gradient à zéro), plus cette amplitude va mécaniquement diminuer.
