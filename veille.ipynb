# Recherches sur le Machine Learning
## A. La science des données

**La science des données** ou **Data Science** est l'ensemble des disciplines cherchant à *donner du sens* aux données.
On y retrouve la **collecte** des **données**, leur **stockage**, leur **formatage**, leur **analyse** et leur **visualisation**.
Elle est au centre de 3 domaines : 
- **Les Mathématiques**
- **L'Expertise métier**
- **L'Informatique**

La Data Science s'oppose à la prise de décisions basée sur l'intuition et permet de prendre des décisions basées sur des données.
Elle nécessite presque toujours une expérience métier lorsqu'elle est mise en application.
On l'associe au Big Data, au Machine Learning et existe "officiellement" depuis l'explosion des bases de données virtuelles avec Internet, fin XXème siècle.

## B. L'apprentissage automatique (Machine Learning)

L'apprentissage automatique est une technique qui permet de donner la compétence d'"*apprendre*" aux machines.
C'est le pan de l'intelligence artificielle le plus recherché et prisé de notre époque.<br> Il contient les "***réseaux de neurones***" et le "***deep learning***".<br><br>
On y voit des "***modèles***" de prédiction, qui vont être capable d'**ajuster** différents paramètres suite à des **observations**. Ce qui donne l'impression que ces modèles apprennent d'eux-même grâce à leurs expériences.
<br><br> Il est généralement constitué de 2 phases, 
- La phase d'apprentissage aussi appelée d'entraînement.
- La phase de validation

L'apprentissage profond est une forme d'apprentissage automatique qui va permettre au modèle de fusionner une partie du processus de l'apprentissage automatique.

## C. L'apprentissage supervisé

**L'apprentissage supervisé** est une forme d'apprentissage automatique qui permet de développer un modèle de prédiction. <br>
Les données d'apprentissage de ce modèle sont déjà ***annotées***.<br><br>
Le but de cet apprentissage est de, par la suite, prédire des résultats sur des données non observées lors de la phase d'apprentissage.<br>
L'algorithme utilisée va être modifiée, optimisée afin de réduire la différence entre les prédictions et les sorties attendues.
<br><br>Dans les types de modèles, on va avoir 2 grandes familles :
- Les modèles de classification qui vont classer des variables qualitatives.
- Les modèles de régression qui vont prédire des variables quantitatives.

## D. L'apprentissage non-supervisé

L'apprentissage non-supervisé se place en opposition à l'apprentissage supervisé par rapport aux données d'entraînement.<br>
Les données d'entraînement ne sont pas ***annotées***.<br>
L'algorithme du modèle va devoir déterminer des *structures sous-jacentes* aux données d'entraînement. Il devra trouver des ***classes ou des groupes*** présentant les mêmes caractéristiques communes.<br><br>
Dans ce cas, le **nombre de classes** *ne sont pas* données en amont et le modèle va essayer de réduire les erreurs qu'il fait avec des **algorithmes d'optimisation**.<br><br>
Il va ainsi altérer le processus de prédiction avec différents paramètres, ici appelés "***poids***".

### E. La classification supervisée

- Le classement automatique ou **classification supervisée** sert à catégoriser des données étiquetées à partir d'algorithme et/ou réseau de neurones. </br>
- En se basant sur des données statistiques, il attribue une classe à un objet/humain. </br>
- La classification supervisée se fait grâce à l'apprentissage supervisée.

**Son objectif =** Prédire, classer, catégoriser les données étiquetées </br>
**Type de donnée utilisée =** Données étiquetées. </br>
**Exemple de cas concret =** Plusieurs images d'animaux, avec une étiquette donnant l'espèce de chaque animal constituent sa base de données. Puis en donnant une nouvelle image, il doit deviner l'espèce de celui-ci. </br>

### F. La classification non supervisée

- La **classification non supervisée** utilise des données non étiquetées pour s'entraîner (apprentissage non supervisée) et classifie les données en fonction de ce qu'il a compris. </br>
- Elle fonctionne indépendamment pour apprendre la structure naturel des données, sans conseils ni instructions spécifiques. </br>
- Il suffit de fournir les données d'entrée et de laisser l'algorithme identifier tout modèle déjà présent dans l'ensemble de données. </br>

**Son objectif =** Organiser, segmenter ou simplifier les données, sans connaissance a priori. </br>
**Type de donnée utilisée =** Données non étiquetées. </br>
**Exemple de cas concret =** Pleins de fruits et légumes composent sa base de donnée, et en se basant sur son algorithme il va deviner, sans connaissance au préalable, comment catégoriser tous ces aliments. </br>

### G. La régression

- Méthode statistique utilisée pour trouver la relation entre les variables et qui permet de prédire des résultats en fonction d'un ensemble de variables d'entrée.
- Dans un contexte de ML, la régression linéaire trouve la relation entre les caractéristiques et un libellé.
- Elle est matérialisée par un graphique et se base sur une table de donnée.

**Son objectif =** Prévoir la valeur de données inconnues en utilisant une autre valeur de données apparentée et connue. </br>
**Type de donnée utilisée =** Nombres/Chiffres</br>

### H. La validation croisée

- Méthodes de rééchantillonnage et de division d'échantillons qui utilisent différentes portions des données pour tester et entraîner un modèle.
- Elle est souvent employée dans des contextes où l'objectif est la prédiction, et où l'on souhaite estimer la précision pratique d'un modèle prédictif.
- Il est nécessaire d’écarter en amont une partie des données du dataset d’entraînement. Ces données seront utilisées pour tester et valider le modèle uniquement.
- Les données : + variable/diversité = plus le modèle apprend mieux à fonctionner.

**Son objectif =** Tester les performances et la précision d'un modèle prédictif de ML à généraliser. Régler les hyperparamètres d'un modèle. Comparer plusieurs modèles pour sélectionner celui le plus approprié.</br>
**Type de donnée utilisée =** Modèles entraînés savec des données étiquetées ainsi que non étiquetées.</br>

*A savoir : Si le modèle est peu performant sur différents ensembles de données de test mais performant sur les données d'entraînement lors des tests, cela indique un surapprentissage.*

## I. Les données d’entraînement, de test et/ou de validation

### Pour entraîner un modèle de machine learning sur un jeu de données, il faut s'assurer que notre modèle n'apprend pas les résultats par cœur.
Les résultats de test sur des exemples différents sont plus solides que des preuves sur un seul ensemble de données.
Pour obtenir ces différents exemples, on sépare le dataset en deux (souvent 80/20) :

- Un ensemble d'entraînement pour entraîner le modèle
- Un ensemble de test pour évaluer le modèle

![training set .png](<attachment:training set .png>)

### Cette approche n'est pas optimale. Supposons que l'on entraîne un modèle sur le "training set" et que l'on évalue sur l'ensemble test à plusieurs reprises.
À chaque tour, si on utilise l'ensemble de test pour ajuster les hyperparamètres et les fonctionnalités,
le modèle risque de s'adapter étroitement à l'ensemble de test (surapprentissage).

![training set avec validation .png](<attachment:training set avec validation .png>)

Il faut ajouter un ensemble de validation qui sera utilisé pour évaluer l'ensemble d'entraînement.
Une fois l'ensemble d'entraînement validé par le validation set, on peut utiliser le test set pour vérifier le modèle.

![workflow trainning set .png](<attachment:workflow trainning set .png>)

Un bon ensemble d'entraînement :
- Suffisamment de données pour générer des résultats de test statistiquement pertinents
- Représentatif de l'ensemble de données global. Autrement dit, ne choisissez pas un ensemble de test dont les caractéristiques sont différentes de celles de l'ensemble d'entraînement.
- Représentatif des données réelles que le modèle rencontrera une fois validé
- Aucun exemple en double dans le modèle d'entraînement

## J. Corrélation linéaire (de Pearson) entre deux variables

### La corrélation de Pearson examine la relation entre deux variables.

### Force de corrélation

0,0 < 0,1 : Pas de corrélation
0,1 < 0,3 : Faible corrélation
0,3 < 0,5 : Corrélation moyenne
0,5 < 0,7 : Corrélation élevée
0,7 < 1 : Corrélation très élevée

Une corrélation est dite positive lorsque les valeurs d'une variable sont associées aux valeurs d'une autre variable, donc les deux évoluent en même temps et dans le même sens.
Une corrélation est dite négative lorsque les valeurs d'une variable évoluent en même temps mais dans des sens opposés.

### Calculer la corrélation de Pearson

### Hypothèses dans l'analyse de corrélation de Pearson

*Hypothèse nulle* : Le coefficient de corrélation de Pearson n'est pas significativement différent de 0 → il n'y a pas de relation linéaire.
*Hypothèse alternative* : Le coefficient de corrélation s'écarte significativement de 0 → il existe une corrélation linéaire.

### Significativité et test t

La question de savoir si le coefficient de corrélation de Pearson est significativement différent de zéro sur la base de l'échantillon étudié peut être vérifiée à l'aide d'un test t. Ici, r est le coefficient de corrélation et n la taille de l'échantillon.

Une valeur p peut alors être calculée à partir de la statistique de test t. Si la valeur p est inférieure au seuil de signification spécifié, généralement 5 %, l'hypothèse nulle est rejetée, sinon elle ne l'est pas.

#### Conditions de la corrélation de Pearson
Mais qu'en est-il des prérequis pour une corrélation de Pearson ? Il faut distinguer si l'on veut simplement calculer le coefficient de corrélation de Pearson ou tester une hypothèse.

Pour calculer le coefficient de corrélation de Pearson, il faut seulement deux variables métriques. Les variables métriques sont, par exemple, le poids d'une personne, son salaire ou sa consommation d'électricité.

Le coefficient de corrélation de Pearson nous indique alors l'importance de la relation linéaire. S'il existe une corrélation non linéaire, nous ne pouvons pas la déduire du coefficient de corrélation de Pearson.

## K. Une fonction de coût

#### La fonction de coût permet de mesurer l'écart entre les prédictions d'un modèle et les valeurs réelles.

### L'objectif est donc de minimiser cette fonction de coût.

Il existe plusieurs fonctions de coût.
Chacune répond à un problème spécifique et aux caractéristiques des données.

### Les principales fonctions de coût sont :
- La moyenne des erreurs quadratiques (MSE)
- L'erreur absolue moyenne (MAE)
- Log Loss (Logarithme de perte)

#### La moyenne des erreurs quadratiques (MSE)

Elle est principalement utilisée pour des modèles de régression.
Cette méthode des moindres carrés pénalise fortement les grandes erreurs en les élevant au carré.
**Utile pour des prédictions où les écarts importants sont critiques.**

![F MSE.png](<attachment:F MSE.png>)

J(θ) : la valeur de la fonction de coût (ou « erreur moyenne quadratique ») pour un ensemble donné de paramètres θ

m : le nombre total d’exemples dans le dataset.
hθ(x(i)) : la prédiction du modèle pour l’exemple i, souvent appelée « hypothèse ».

y(i) : la valeur réelle attendue pour l’exemple i.

(hθ(x(i))−y(i))² : le carré de l’erreur entre la prédiction et la vraie valeur pour l’exemple i.

#### L'erreur absolue moyenne (MAE)

Il s’agit d’une autre formule pour les modèles de régression linéaire. Mais à la différence de la précédente, cette fonction de coût mesure la moyenne des erreurs absolues entre les valeurs réelles et les prédictions d’un modèle. Et comme la MAE n’élève pas les erreurs au carré, elle est beaucoup plus robuste face aux valeurs aberrantes.

### Log Loss (algorithme de perte)

Cette fonction de coût est utilisée pour évaluer les modèles de classification probabiliste, notamment dans les problèmes de classification binaire ou multiclasse.

Le Log Loss mesure la distance entre les distributions des probabilités prédites et les vraies classes.
